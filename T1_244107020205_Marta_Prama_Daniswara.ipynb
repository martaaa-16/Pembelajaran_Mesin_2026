```python
import pkgutil
import importlib.machinery

# Tambahkan metode find_module agar kompatibel dengan pkg_resources lama
if not hasattr(importlib.machinery.FileFinder, 'find_module'):
    def _find_module(self, fullname):
        spec = self.find_spec(fullname)
        return spec.loader if spec else None
    importlib.machinery.FileFinder.find_module = _find_module

if not hasattr(pkgutil, 'ImpImporter'):
    pkgutil.ImpImporter = importlib.machinery.FileFinder

import pyecg
print("Versi pyECG berhasil dimuat:", pyecg.__version__)
```
